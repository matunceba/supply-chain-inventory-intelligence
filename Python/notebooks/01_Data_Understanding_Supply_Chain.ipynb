### Replenishment Logic and Reorder Point Assessment

The relationship between previous inventory levels and order placement was investigated using 91,000 observations. The first observation for each SKU–warehouse combination was excluded because no previous inventory value was available.

The analysis identified the following pattern:

- All 85,973 observations where previous inventory was above the reorder point had no recorded order quantity.
- All 254 observations where previous inventory was exactly at the reorder point had a positive order quantity.
- All 4,773 observations where previous inventory was below the reorder point had a positive order quantity.

Consequently, all 5,027 positive order records occurred when previous inventory was at or below the reorder point.

A separate consistency assessment found that none of the 250 SKU–warehouse combinations had a varying reorder point.

**Interpretation:** The dataset exhibits a deterministic replenishment pattern: positive order quantities are recorded when previous inventory reaches or falls below the reorder point. This relationship is consistent with the inventory reconciliation results.

**Limitation:** These findings describe the structure and relationships within the available dataset. They do not independently establish that the same replenishment policy would apply in a real-world supply chain operation.

```python
# Create a separate DataFrame for inventory movement analysis.
df_order_analysis = df_working.copy()

# Identify the previous inventory level within each SKU-warehouse group.
df_order_analysis["Previous_Inventory"] = (
    df_order_analysis
    .groupby(["SKU_ID", "Warehouse_ID"])["Inventory_Level"]
    .shift(1)
)

# Identify the next inventory level within each SKU-warehouse group.
df_order_analysis["Next_Inventory"] = (
    df_order_analysis
    .groupby(["SKU_ID", "Warehouse_ID"])["Inventory_Level"]
    .shift(-1)
)

# Identify the previous observation date within each group.
df_order_analysis["Previous_Date"] = (
    df_order_analysis
    .groupby(["SKU_ID", "Warehouse_ID"])["Date"]
    .shift(1)
)

# Identify the next observation date within each group.
df_order_analysis["Next_Date"] = (
    df_order_analysis
    .groupby(["SKU_ID", "Warehouse_ID"])["Date"]
    .shift(-1)
)

# Display selected columns to verify the new variables.
display(
    df_order_analysis[
        [
            "Date",
            "SKU_ID",
            "Warehouse_ID",
            "Inventory_Level",
            "Previous_Inventory",
            "Next_Inventory",
            "Previous_Date",
            "Next_Date"
        ]
    ].head(10)
)
```

```python
# Create a separate copy for the date continuity assessment.
date_continuity = df_order_analysis.copy()

# Calculate the number of calendar days between consecutive observations.
date_continuity["Date_Gap_Days"] = (
    date_continuity["Date"] -
    date_continuity["Previous_Date"]
).dt.days

# Exclude the first observation in each SKU-warehouse group,
# because it has no previous observation.
date_continuity = date_continuity.dropna(
    subset=["Previous_Date"]
).copy()

# Count the number of observations for each date-gap value.
date_gap_summary = (
    date_continuity["Date_Gap_Days"]
    .value_counts()
    .sort_index()
    .rename_axis("Days_Between_Records")
    .to_frame("Observation_Count")
)

# Display the date-gap distribution.
display(date_gap_summary)

# Identify records where the gap is not exactly one calendar day.
non_daily_gaps = date_continuity[
    date_continuity["Date_Gap_Days"] != 1
].copy()

# Report the continuity assessment results.
print("TOTAL CONSECUTIVE OBSERVATIONS:", len(date_continuity))

print(
    "OBSERVATIONS WITH A ONE-DAY GAP:",
    (date_continuity["Date_Gap_Days"] == 1).sum()
)

print("OBSERVATIONS WITH OTHER DATE GAPS:", len(non_daily_gaps))

# Display examples of any date gaps requiring investigation.
display(
    non_daily_gaps[
        [
            "SKU_ID",
            "Warehouse_ID",
            "Previous_Date",
            "Date",
            "Date_Gap_Days"
        ]
    ].head(20)
)
```

### Date Continuity and Chronological Inventory Validation

A date-continuity assessment was conducted to determine whether consecutive records within each SKU-warehouse combination represented consecutive calendar days.

The dataset contains 91,250 observations distributed across 250 SKU-warehouse combinations. The first observation in each combination was excluded because no preceding inventory observation was available, leaving 91,000 consecutive observations for validation.

**Results:**

| Validation measure | Result |
|---|---:|
| Consecutive observations assessed | 91,000 |
| Observations with a one-day gap | 91,000 |
| Observations with other date gaps | 0 |
| Date continuity rate | 100% |

All assessed observations had a one-day interval between consecutive records.

**Interpretation:**

The results confirm daily continuity within each SKU-warehouse combination over the recorded observation period. Consequently, the `Previous_Inventory` and `Next_Inventory` variables represent inventory levels from the immediately preceding and following calendar days, respectively.

This validation supports the chronological inventory movement analysis and the subsequent inventory reconciliation assessment.

**Limitation:** The validation establishes continuity within the available dataset. It does not independently establish the real-world operational meaning of every inventory-related variable.
